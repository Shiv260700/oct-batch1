# 02 · A local language model
- **No API, no Ollama**: a small open-source model (Qwen2.5-0.5B-Instruct, Apache-2.0) runs on this laptop with `transformers`
- Downloaded once (~1 GB) into the Hugging Face cache; after that it works offline
- Set `MODEL_ID=Qwen/Qwen2.5-1.5B-Instruct` for better answers (bigger, slower)

```python
import os
os.environ['HF_HUB_OFFLINE'] = '1'          # models come from the local cache only
os.environ['TOKENIZERS_PARALLELISM'] = 'false'
import warnings; warnings.filterwarnings('ignore')
import pandas as pd
pd.set_option('display.max_colwidth', 120)
from logbot.llm import LocalLLM
llm = LocalLLM()
print(llm.model_id, 'on', llm.device)
```

## First answer

```python
reply, stats = llm.chat([{'role': 'user', 'content': 'In one sentence: what does a 502 Bad Gateway from Apache mean?'}])
print(reply); stats
```

## Without the logs it can only guess
- Ask about *our* shop: it has never seen our logs, so it invents an answer (a **hallucination**)

```python
reply, _ = llm.chat([{'role': 'user', 'content': 'Why did shop-app return 500 errors on /api/orders on 13 October 2026?'}])
print(reply)
```

```python
assert len(reply) > 20 and stats['tokens_per_s'] > 1
print(f"✓ model runs locally: {stats['tokens_per_s']} tokens/s; next: give it the logs")
```
