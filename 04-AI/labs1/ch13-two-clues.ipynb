# Chapter 13 · Two clues at once
**Deck 3 · Live demo.** Learn w₁ (per hour), w₂ (per practice test) and b from four students, then predict Asha.

```python
import numpy as np
import matplotlib.pyplot as plt
np.set_printoptions(precision=3, suppress=True)
plt.rcParams.update({"figure.figsize": (8, 4.5), "axes.grid": True, "font.size": 12})
```

## 1 · Your prediction
E: 2 h, 1 test → 46 · F: 4 h, 1 test → 56 · G: 2 h, 3 tests → 54 · H: 4 h, 3 tests → 64. **Asha: 8 hours, 2 tests?**

```python
my_prediction = 80
```

```python
X = np.array([[2, 1], [4, 1], [2, 3], [4, 3]], dtype=float)   # columns: hours, tests
y = np.array([46, 56, 54, 64], dtype=float)
asha = np.array([8, 2], dtype=float)
```

## 2 · The model
$$\hat{y} = w_1 x_1 + w_2 x_2 + b = \mathbf{w}\cdot\mathbf{x} + b$$ A weighted sum is a receipt: clues are quantities, weights are prices. `X @ w` does every student at once.

## 3 · Gradient descent with three knobs
One slope per knob: errors × hours for w₁, errors × tests for w₂, errors alone for b.

```python
w = np.zeros(2); b = 0.0; eta = 0.04
for step in range(10_000):
    errors = X @ w + b - y
    w -= eta * 2 * X.T @ errors / len(y)
    b -= eta * 2 * errors.mean()
print('w1, w2 =', w.round(3), ' b =', round(b, 3))
print('Asha →', round(asha @ w + b, 2))
```

## 4 · PROOF

```python
assert np.allclose(w, [5, 4], atol=1e-3) and abs(b - 32) < 1e-3, 'w1 = 5, w2 = 4, b = 32'
assert np.allclose(X @ [5, 4] + 32, y), 'the rule reproduces every student'
assert round(asha @ w + b, 2) == 80 == my_prediction, 'Asha → 80'
print('✓ proof: 5, 4, 32 → Asha 80')
```

## 5 · The same answer from scikit-learn

```python
from sklearn.linear_model import LinearRegression
m = LinearRegression().fit(X, y)
print('sklearn:', m.coef_.round(3), round(m.intercept_, 3), '→ Asha', m.predict([asha]).round(2))
```
