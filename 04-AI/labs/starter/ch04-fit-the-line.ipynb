## 5b · Unroll the sigma
The book notation $L = \frac{1}{n}\sum_{i=1}^{n} (\hat{y}_i - y_i)^2$ is just one term per student, then add, then divide by n.

```python
guess = predict(2, 55, hours)
terms = [(guess[i] - marks[i]) ** 2 for i in range(3)]   # i = 1, 2, 3 in the slides
print('terms :', terms, '→ Σ =', sum(terms), '→ L = Σ / n =', sum(terms) / len(terms))
```
